```python
import pandas as pd

# Hiển thị đầy đủ cột cho dataset nhỏ (101 dòng x 11 cột)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 60)

# CHỈ ĐỌC — không ghi đè file gốc
RAW_PATH = "../data/raw/Student Mental health.csv"
df = pd.read_csv(RAW_PATH)
print("Đã đọc dữ liệu từ:", RAW_PATH)
```

## 3. Giá trị thiếu (missing values)

```python
missing = df.isnull().sum()
print("Số lượng giá trị thiếu theo từng cột:")
print(missing)
print("\nTổng số ô thiếu trong toàn bộ dữ liệu:", int(missing.sum()))
```

```python
rows_with_missing = df[df.isnull().any(axis=1)]
print(f"Số dòng có chứa ít nhất một giá trị thiếu: {len(rows_with_missing)}")
rows_with_missing
```

## 4. Dòng trùng lặp hoàn toàn (duplicated rows)

```python
n_dup = int(df.duplicated().sum())
print(f"Số dòng trùng lặp hoàn toàn: {n_dup}")
```

## 5. Phân bố biến mục tiêu `Do you have Depression?`

```python
target = "Do you have Depression?"

counts = df[target].value_counts(dropna=False)
pct = df[target].value_counts(dropna=False, normalize=True).mul(100).round(2)

dist = pd.DataFrame({"count": counts, "percentage_%": pct})
print(dist)
```

## 6. Khoảng trắng thừa và giá trị duy nhất (unique values)

```python
print("6a. Cột có khoảng trắng thừa đầu/cuối:")
has_ws = [c for c in df.columns if c != c.strip()]
if has_ws:
    for c in has_ws:
        print(f"    {c!r} -> {c.strip()!r}")
else:
    print("    (không có)")
```

```python
print("6b. Giá trị của cột dạng object có khoảng trắng thừa đầu/cuối:")
object_cols = df.select_dtypes(include="object").columns
found = False
for col in object_cols:
    series = df[col].dropna().astype(str)
    mask = series.str.strip() != series
    if mask.any():
        found = True
        print(f"\n[{col}] ({int(mask.sum())} giá trị bị thừa khoảng trắng):")
        print(series[mask].value_counts().to_string())
if not found:
    print("    (không có)")
```

```python
print("6c. Toàn bộ giá trị duy nhất của từng cột (trừ Timestamp):")
for col in df.columns:
    if col == "Timestamp":
        continue
    uniq = df[col].dropna().unique()
    print(f"\n{col!r}  ({len(uniq)} giá trị duy nhất):")
    for v in sorted(uniq, key=str):
        print(f"    {v!r}")
```
